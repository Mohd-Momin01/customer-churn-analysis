```python
import pandas as pd
import numpy as np
import random
from faker import Faker

# Reproducibility
np.random.seed(42)
random.seed(42)
Faker.seed(42)

fake = Faker("en_IN")

# Start with 10,000 while testing
N = 10_000

TODAY = pd.Timestamp("2026-08-10")

print("Libraries loaded successfully.")
print("Rows to generate:", N)
```

```python
customer_id = [
    f"C{str(i).zfill(6)}"
    for i in range(1, N + 1)
]

customer_name = [
    fake.name()
    for _ in range(N)
]

print("Customer IDs and names generated.")
print(customer_id[:5])
print(customer_name[:5])
```
