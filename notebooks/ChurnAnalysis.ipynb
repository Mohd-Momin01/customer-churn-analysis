```python
import pandas as pd
import numpy as np
import random
from faker import Faker

np.random.seed(42)
random.seed(42)
Faker.seed(42)

fake = Faker("en_IN")

N = 10_000

TODAY = pd.Timestamp("2026-08-10")

print("Libraries loaded successfully.")
print("Rows to generate:", N)
```

```python
customer_id = [
    f"C{str(i).zfill(6)}"
    for i in range(1, N + 1)
]

customer_name = [
    fake.name()
    for _ in range(N)
]

print("Customer IDs and names generated.")
print(customer_id[:5])
print(customer_name[:5])
```

```python
gender = np.random.choice(
    ["Male", "Female"],
    N,
    p=[0.55, 0.45]
)

age = np.random.randint(
    18,
    71,
    N
)

senior_citizen = np.where(
    age >= 60,
    "Yes",
    "No"
)

dependents = np.random.choice(
    ["Yes", "No"],
    N,
    p=[0.38, 0.62]
)

print("Demographic data generated.")
```

```python
locations = {
    "Maharashtra": ["Mumbai", "Pune", "Nagpur", "Nashik"],
    "Delhi": ["New Delhi", "Delhi"],
    "Karnataka": ["Bengaluru", "Mysuru", "Mangaluru"],
    "Telangana": ["Hyderabad", "Warangal"],
    "Tamil Nadu": ["Chennai", "Coimbatore", "Madurai"],
    "Gujarat": ["Ahmedabad", "Surat", "Vadodara"],
    "West Bengal": ["Kolkata", "Howrah"],
    "Rajasthan": ["Jaipur", "Jodhpur", "Udaipur"],
    "Uttar Pradesh": ["Lucknow", "Noida", "Kanpur", "Agra"],
    "Madhya Pradesh": ["Indore", "Bhopal"],
    "Punjab": ["Chandigarh", "Ludhiana", "Amritsar"],
    "Kerala": ["Kochi", "Thiruvananthapuram"]
}

states = list(locations.keys())

state = np.random.choice(
    states,
    N
)

city = [
    random.choice(locations[s])
    for s in state
]

# Realistic-looking Indian PIN codes
pin_code = [
    fake.postcode()
    for _ in range(N)
]

print("Location data generated.")
```

```python
customer_segment = np.random.choice(
    [
        "Mass Market",
        "Value",
        "Premium",
        "High Value"
    ],
    N,
    p=[0.40, 0.30, 0.20, 0.10]
)

print("Customer segments generated.")
```

```python
tenure_months = np.random.randint(
    1,
    73,
    N
)

subscription_type = np.random.choice(
    [
        "Basic",
        "Standard",
        "Premium"
    ],
    N,
    p=[0.40, 0.40, 0.20]
)

contract_type = np.random.choice(
    [
        "Month-to-Month",
        "One Year",
        "Two Year"
    ],
    N,
    p=[0.55, 0.25, 0.20]
)

print("Subscription data generated.")
```

```python
internet_service = np.random.choice(
    [
        "DSL",
        "Fiber",
        "5G",
        "Cable",
        "None"
    ],
    N,
    p=[0.20, 0.35, 0.20, 0.20, 0.05]
)

tech_support = np.random.choice(
    ["Yes", "No"],
    N,
    p=[0.60, 0.40]
)

print("Internet and technical support data generated.")
```

```python
monthly_charges = np.zeros(N)

for i in range(N):

    if subscription_type[i] == "Basic":
        charge = np.random.normal(399, 60)

    elif subscription_type[i] == "Standard":
        charge = np.random.normal(699, 90)

    else:
        charge = np.random.normal(1199, 150)

    if internet_service[i] == "Fiber":
        charge += 150

    elif internet_service[i] == "5G":
        charge += 100

    monthly_charges[i] = max(
        charge,
        199
    )

monthly_charges = np.round(
    monthly_charges,
    2
)

discount_percent = np.random.choice(
    [0, 5, 10, 15, 20],
    N,
    p=[0.40, 0.25, 0.20, 0.10, 0.05]
)

total_charges = np.round(
    monthly_charges
    * tenure_months
    * (1 - discount_percent / 100),
    2
)

lifetime_value = np.round(
    total_charges +
    monthly_charges * np.random.uniform(2, 8, N),
    2
)

print("Financial data generated.")
```

```python
payment_method = np.random.choice(
    [
        "UPI",
        "Credit Card",
        "Debit Card",
        "Net Banking",
        "Wallet"
    ],
    N,
    p=[0.30, 0.25, 0.20, 0.15, 0.10]
)

auto_pay = np.random.choice(
    ["Yes", "No"],
    N,
    p=[0.62, 0.38]
)

payment_delay_days = np.random.poisson(
    2,
    N
)

payment_delay_days = np.clip(
    payment_delay_days,
    0,
    30
)

print("Payment data generated.")
```

```python
monthly_data_gb = np.round(
    np.random.gamma(
        shape=4,
        scale=12,
        size=N
    ),
    2
)

avg_call_minutes = np.round(
    np.random.gamma(
        shape=4,
        scale=35,
        size=N
    ),
    0
)

login_frequency = np.random.poisson(
    12,
    N
)

referral_count = np.random.poisson(
    1.2,
    N
)

print("Customer behaviour data generated.")
```

```python
support_tickets = np.random.poisson(
    2.5,
    N
)

complaints = np.random.poisson(
    0.8,
    N
)

satisfaction_score = np.random.randint(
    1,
    11,
    N
)

nps_score = np.random.randint(
    0,
    11,
    N
)

print("Support and satisfaction data generated.")
```

```python
# Last Interaction Date

days_ago = np.random.randint(
    0,
    91,
    N
)

last_interaction_date = (
    TODAY - pd.to_timedelta(
        days_ago,
        unit="D"
    )
)

print("Interaction dates generated.")
print(last_interaction_date[:5])
```

```python
acquisition_channel = np.random.choice(
    [
        "Website",
        "Google Ads",
        "Social Media",
        "Referral",
        "Retail Store",
        "Partner",
        "Telemarketing"
    ],
    N,
    p=[
        0.20,
        0.15,
        0.15,
        0.15,
        0.15,
        0.10,
        0.10
    ]
)

offer_received = np.random.choice(
    [
        "No Offer",
        "10% Discount",
        "20% Discount",
        "Free Upgrade",
        "Extra Data"
    ],
    N,
    p=[
        0.45,
        0.20,
        0.15,
        0.10,
        0.10
    ]
)

print("Marketing data generated.")
```

```python
churn_score = (
    0.08

    # Contract
    + (contract_type == "Month-to-Month") * 0.16
    - (contract_type == "Two Year") * 0.12

    # Tenure
    + (tenure_months < 12) * 0.12
    - (tenure_months >= 36) * 0.08

    # Customer support
    + (support_tickets >= 5) * 0.10
    + (complaints >= 2) * 0.12

    # Satisfaction
    + (satisfaction_score <= 4) * 0.18
    - (satisfaction_score >= 8) * 0.08

    # Payment
    + (payment_delay_days >= 7) * 0.10
    - (auto_pay == "Yes") * 0.05

    # Engagement
    + (login_frequency <= 3) * 0.10
)

# Random variation
churn_score += np.random.normal(
    0,
    0.05,
    N
)

churn_probability = np.clip(
    churn_score,
    0.02,
    0.90
)

churn = np.random.binomial(
    1,
    churn_probability
)

churn_status = np.where(
    churn == 1,
    "Yes",
    "No"
)

print("Churn status generated.")
```

```python
churn_reason = []

for i in range(N):

    if churn_status[i] == "No":
        reason = "Not Churned"

    elif satisfaction_score[i] <= 3:
        reason = "Poor Service"

    elif complaints[i] >= 2:
        reason = "Customer Support Issues"

    elif payment_delay_days[i] >= 7:
        reason = "Payment Issues"

    elif contract_type[i] == "Month-to-Month":
        reason = "Contract Flexibility"

    else:
        reason = "Competitor Offer"

    churn_reason.append(reason)

print("Churn reasons generated.")
```

```python
df = pd.DataFrame({

    "Customer_ID": customer_id,
    "Customer_Name": customer_name,

    "Gender": gender,
    "Age": age,
    "Senior_Citizen": senior_citizen,
    "Dependents": dependents,

    "State": state,
    "City": city,
    "PIN_Code": pin_code,

    "Customer_Segment": customer_segment,

    "Tenure_Months": tenure_months,
    "Subscription_Type": subscription_type,
    "Contract_Type": contract_type,

    "Internet_Service": internet_service,
    "Tech_Support": tech_support,

    "Monthly_Charges": monthly_charges,
    "Total_Charges": total_charges,
    "Discount_Percent": discount_percent,
    "Lifetime_Value": lifetime_value,

    "Payment_Method": payment_method,
    "Auto_Pay": auto_pay,
    "Payment_Delay_Days": payment_delay_days,

    "Monthly_Data_GB": monthly_data_gb,
    "Avg_Call_Minutes": avg_call_minutes,
    "Login_Frequency": login_frequency,

    "Support_Tickets": support_tickets,
    "Complaints": complaints,

    "Satisfaction_Score": satisfaction_score,
    "NPS_Score": nps_score,

    "Referral_Count": referral_count,

    "Acquisition_Channel": acquisition_channel,
    "Offer_Received": offer_received,

    "Last_Interaction_Date": last_interaction_date,

    "Churn": churn_status,
    "Churn_Reason": churn_reason
})

print("Final dataset created!")
print("Shape:", df.shape)

df.head()
```
