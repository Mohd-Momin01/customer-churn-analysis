```python
churn_rate = df["Churn_Flag"].mean() * 100

print(f"Overall Churn Rate: {churn_rate:.2f}%")
```

```python
# Churn by Subscription
contract_churn = (
    df.groupby("Contract_Type")
      .agg(
          Customers=("Customer_ID", "count"),
          Churned=("Churn_Flag", "sum"),
          Churn_Rate=("Churn_Flag", "mean")
      )
      .reset_index()
)

contract_churn["Churn_Rate"] *= 100

contract_churn
```

```python
#Churn by Customer Segment
subscription_churn = (
    df.groupby("Subscription_Type")
      .agg(
          Customers=("Customer_ID", "count"),
          Churned=("Churn_Flag", "sum"),
          Churn_Rate=("Churn_Flag", "mean")
      )
      .reset_index()
)

subscription_churn["Churn_Rate"] *= 100

subscription_churn
```

```python
 # Churn by Internet Service
internet_churn = (
    df.groupby("Internet_Service")
      .agg(
          Customers=("Customer_ID", "count"),
          Churned=("Churn_Flag", "sum"),
          Churn_Rate=("Churn_Flag", "mean")
      )
      .reset_index()
)

internet_churn["Churn_Rate"] *= 100

internet_churn
```

```python
# Churn by Age Group
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 55, 65, 100],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56-65",
        "65+"
    ]
)

age_churn = (
    df.groupby("Age_Group", observed=True)["Churn_Flag"]
      .mean()
      .mul(100)
      .reset_index(name="Churn_Rate")
)

age_churn
```

```python
# Revenue at Risk by Churn Status
revenue_risk = (
    df.groupby("Churn")
      .agg(
          Customers=("Customer_ID", "count"),
          Monthly_Revenue=("Monthly_Charges", "sum"),
          Estimated_LTV=("Estimated_LTV", "sum")
      )
      .reset_index()
)

revenue_risk

churned_revenue = df.loc[
    df["Churn_Flag"] == 1,
    "Monthly_Charges"
].sum()

print(
    f"Monthly revenue associated with churned customers: "
    f"{churned_revenue:,.2f}"
)
```

```python
df["Risk_Segment"] = np.select(
    [
        (df["Churn_Flag"] == 1),
        (df["Tenure_Months"] < 12) & (df["Monthly_Charges"] > df["Monthly_Charges"].median()),
        (df["Days_Since_Last_Interaction"] > 60)
    ],
    [
        "Churned",
        "High Risk",
        "Inactive Risk"
    ],
    default="Low/Normal Risk"
)

df["Risk_Segment"].value_counts()
```
